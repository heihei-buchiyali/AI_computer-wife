```python
import pandas as pd

pd.set_option('display.min_rows',30)#设置显示行数（默认十行 ）


#1、读取数据
df=pd.read_csv('data/sales.csv')

#2、数据清洗
#2.1、缺失值处理
df.isnull().sum()#查看缺失值数量

#2.1.1删除缺失值
df.dropna()#删除缺失值所在行
df.dropna(axis=1)#删除缺失值所在列
#2.1.1填充缺失值
df.fillna(0)#用0填充缺失值
df.fillna(method='ffill')#用上一行填充缺失值
df.fillna(method='bfill')#用下一行填充缺失值
df.fillna(df.mean())#用均值填充缺失值
df.fillna(df.median())#用中位数填充缺失值
df.fillna(df.mode().iloc[0])#用众数填充缺失值



```

```python
#2.2重复值的处理
#2.2.1查看重复值
df.duplicated().sum()#查看重复值数量
#2.2.2删除重复值
df.drop_duplicates()#删除重复值

```

```python
#2.3异常处理
#2.3.1查看异常值
df.describe()#查看异常值

#2.3.2删除异常值
df.drop(df[df['列名']条件].index)

#2.3.3修复异常值
df['列名'].replace(旧值,新值)
```
